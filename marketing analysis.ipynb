```python
import pandas as pd
dt1 = pd.read_csv(r"c:\Users\Admin\Documents\datasets\website_wata.csv")
print(dt1)
```

```python
dt2 = pd.read_csv(r"C:\Users\Admin\Documents\datasets\marketing_campaign_dataset.csv")
print(dt2)
```

```python
dt3 = pd.merge(dt1 , dt2 , on = "Campaign_ID" , how = "inner")
print(dt3)



```

```python
dt3.info()
dt3.head()
dt3.describe()
dt3.isnull().sum()
```

```python
dt3 = dt3.drop_duplicates()
print(dt3)
```

```python
null_values = dt3.isnull().sum()
print(null_values) 
```

```python
datatypes = dt3['Date'] = pd.to_datetime(dt3["Date"])
print(datatypes)

```

```python
numerics1 = dt3["Session Duration  "] = pd.to_numeric(dt3["Session Duration"], errors="coerce")



print(numerics1)
```

```python
dt3['conversions'] = dt3['Clicks'] * dt3['Conversion_Rate']
dt3['Acquisition_Cost'] = pd.to_numeric(
    dt3['Acquisition_Cost'].astype(str).str.replace(r'[^0-9.-]', '', regex=True),
    errors='coerce'
 )
CAC = dt3['Acquisition_Cost'].div(dt3['conversions'].replace(0, pd.NA))
dt3['CAC'] = CAC

dt3[['Acquisition_Cost', 'conversions', 'CAC']].describe()
```

```python
conversions = dt3['Conversion_Rate'] / dt3['Previous Visits']
print("the conversion rate per channel is" ,conversions)
```

```python
ROI = dt3['ROI']
print(ROI)
ROI = dt3['ROI'].sum()
print("the total ROI is", ROI)
```

```python
import matplotlib.pyplot as plt
print("CAC = Previous Visits / Engagement Score")

x = dt3['Previous Visits']
y = dt3['Engagement_Score']
plt.bar(x, y)
plt.xlabel("Previous Visits")
plt.ylabel("Engagement Score")
plt.title("CAC Analysis")
plt.show()
```

```python
print("CAC = Previous Visits / Engagement Score")

x = dt3['Previous Visits']
y = dt3['Engagement_Score']
plt.plot(x, y)
plt.xlabel("Previous Visits")
plt.ylabel("Engagement Score")
plt.title("CAC Analysis")
plt.show()
```

```python
pivot_table = dt3.pivot_table(index='Traffic Source',
                              columns='Target_Audience',
                              values='Conversion_Rate',
                              aggfunc='mean')
print(pivot_table)

pivot_table.plot(kind='bar', stacked=False, figsize=(10, 6))
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
dt3['profit_margin_numeric'] = dt3['profit margin'].eq('Profit').astype(int)
pivot_table = dt3.pivot_table(index='Channel_Used',
                              columns='profit margin',
                              values='ROI',
                              aggfunc='mean')
print(pivot_table)

pivot_table.mul(100).plot(kind='bar', stacked=False, figsize=(10, 6))
plt.ylabel('Profit Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
dt3['Time on Page Band'] = pd.cut(
    dt3['Time on Page'],
    bins=[0, 2, 5, 10, 15, float('inf')],
    labels=['0-2 min', '2-5 min', '5-10 min', '10-15 min', '15+ min'],
    include_lowest=True
 )
pivot_table = dt3.pivot_table(
    index='Traffic Source',
    columns='Time on Page Band',
    values='profit_margin_numeric',
    aggfunc='mean'
 )
print(pivot_table)

pivot_table.mul(100).plot(kind='bar', stacked=True, figsize=(10, 6))
plt.ylabel('Profit Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
